```python
from transformers import pipeline
```

```python
# Available models:
# gpt2
# distilgpt2
# EleutherAI/gpt-neo-125M

model_name = input("Enter model name (gpt2 / distilgpt2 / EleutherAI/gpt-neo-125M): ")
```

Load the Model

```python
print("Loading model...")

generator = pipeline(
    "text-generation",
    model=model_name
)

print("Model Loaded Successfully!")
```

Select Task

```python
print("Choose Task")

print("1. Story Generation")
print("2. Technical Explanation")
print("3. Question & Answer")

choice = input("Enter your choice (1/2/3): ")
```

Enter Prompt

```python
if choice == "1":
    prompt = input("Enter story starting line: ")

elif choice == "2":
    topic = input("Enter technical topic: ")
    prompt = f"Explain {topic} in simple terms."

elif choice == "3":
    question = input("Enter your question: ")
    prompt = f"Question: {question}\nAnswer:"

else:
    print("Invalid Choice")
```

Set Generation Parameters

```python
temperature = float(input("Enter temperature (Example: 0.7): "))

max_length = int(input("Enter maximum length (Example: 100): "))

top_k = int(input("Enter top_k (Example: 50): "))

top_p = float(input("Enter top_p (Example: 0.9): "))
```

Generate Text

```python
output = generator(
    prompt,
    max_length=max_length,
    temperature=temperature,
    top_k=top_k,
    top_p=top_p,
    num_return_sequences=1
)
```

Display Output

```python
print("="*60)
print("Generated Text")
print("="*60)

print(output[0]["generated_text"])
```

Parameter Comparison

```python
print("Parameter Summary")

print(f"Model Used      : {model_name}")
print(f"Temperature     : {temperature}")
print(f"Maximum Length  : {max_length}")
print(f"Top-K           : {top_k}")
print(f"Top-P           : {top_p}")
```

Experiment 1

```python
experiment1 = generator(
    prompt,
    max_length=80,
    temperature=0.3,
    top_k=20,
    top_p=0.8
)

print(experiment1[0]["generated_text"])
```

Experiment 2

```python
experiment2 = generator(
    prompt,
    max_length=100,
    temperature=0.7,
    top_k=50,
    top_p=0.9
)

print(experiment2[0]["generated_text"])
```

Experiment 3

```python
experiment3 = generator(
    prompt,
    max_length=150,
    temperature=1.0,
    top_k=100,
    top_p=0.95
)

print(experiment3[0]["generated_text"])
```

Conclusion

1. Lower temperature produces more focused and deterministic text.

2. Higher temperature produces more creative and diverse text.

3. Increasing max_length generates longer responses.

4. Lower top_k limits word selection and makes output more predictable.

5. Higher top_p allows more diverse token selection while maintaining coherence.

6. Hugging Face Transformers makes it easy to generate stories, technical explanations, and Q&A responses using pre-trained language models.
